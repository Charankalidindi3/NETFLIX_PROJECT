### Incremental Data Loading using Autoloader

```python
%sql
create schema netflix_catalog.netflix_schema;
```

```python
checkpoint_location="abfss://silver@mynetflixstorage3.dfs.core.windows.net/checkpoint"
```

```python
df = spark.readStream.format('cloudFiles').option('cloudFiles.format', 'csv').option('cloudFiles.schemaLocation',checkpoint_location).load('abfss://raw@mynetflixstorage3.dfs.core.windows.net/')
```

```python
df.display()
```

```python
df.writeStream.option('checkpointlocation',checkpoint_location)\
    .trigger(processingTime='10 seconds')\
    .start('abfss://bronze@mynetflixstorage3.dfs.core.windows.net/netflix_MovieorSeries_titles')
```
