### SILVER Notebook Lookup  Table

### Parameters


```python
dbutils.widgets.text("sourcefolder","netflix_directors")
dbutils.widgets.text("targetfolder","netflix_directors")
```

### variobles

```python
var_srcf = dbutils.widgets.get("sourcefolder")
var_trgtf = dbutils.widgets.get("targetfolder")
```

```python
df= spark.read.format('csv')\
    .option('header', True)\
    .option('inferSchema', True)\
    .load(f'abfss://bronze@mynetflixstorage3.dfs.core.windows.net/{var_srcf}')
```

```python
df.display()
```

```python
df.write.format("delta")\
    .mode("append")\
    .option("path",f"abfss://silver@mynetflixstorage3.dfs.core.windows.net/{var_trgtf}")\
    .save()
```
